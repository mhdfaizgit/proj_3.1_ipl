```python
q("""
--which bowler takes a wicket most often - best strike rate in balls per wicket?
--bowler ,total ball,total wicket , balls per wicket

SELECT
    bowler,
    SUM(is_legal) AS total_balls,
    SUM(bowler_wicket) AS total_wickets,
    ROUND(
        1.0 * SUM(is_legal) / SUM(bowler_wicket),
        2
    ) AS balls_per_wicket
FROM v_ball
GROUP BY bowler
HAVING total_wickets > 50
ORDER BY balls_per_wicket 
LIMIT 10;

""")
```

```python
q("""
-- which bowler gives away the fewest boundaries?
-- bowler,balls,4and6,boundary_perc = sumof 4and6 /balls * 100
 
SELECT bowler,SUM(is_legal) AS balls,
       SUM(CASE WHEN batter_runs IN(4,6)
            THEN 1
            ELSE 0
            END)AS bound,
       ROUND(100.0*SUM(CASE WHEN batter_runs IN(4,6)
                           THEN 1 ELSE 0 END )/SUM(is_legal) ,2) AS boundary_perc
FROM v_ball
GROUP BY bowler
having balls>1000
ORDER BY boundary_perc ;

""")
```

```python
q("""
--which bowling style is not economical of all the styles recorded ?

SELECT CASE WHEN bowler_type LIKE                       -- spin or pace 
       '%Legbreak%' OR bowler_type LIKE 
       '%Offbreak%' OR bowler_type LIKE 
       '%Orthodox%' OR bowler_type LIKE 
       '%Wrist spin%' OR bowler_type LIKE 
       '%Googly%' THEN 'Spin' ELSE 'Pace' 
       END AS style,                                    -- branches of th 
       SUM(is_legal) AS balls,                          -- sample size 
       ROUND(6.0*SUM(bowler_runs)/SUM(is_legal),2)      -- runs per over 
              AS economy, 
       ROUND(100.0*SUM(bowler_wicket)/SUM(is_legal),2)  -- bowler's wicke 
              AS wicket_pct 
FROM   v_ball                                           -- one row = one 
WHERE  bowler_type IS NOT NULL AND                      -- drop the blanks 
       TRIM(bowler_type) <> '' 
GROUP  BY style;                                        -- one row per gr


""")
```

```python
q("""
--Do left-arm bowlers trouble right-handed batters more?
--batsman_type,bowler,economy , wicket_perc


SELECT CASE WHEN bowler_type LIKE 'Left%'
        THEN 'Left arm' ELSE 'Right arm' END
     AS arm,
     batsman_type,
     ROUND(6.0*SUM(bowler_runs)/SUM(is_legal),2)     -- runs per over
        AS economy,
     ROUND(100.0*SUM(bowler_wicket)/SUM(is_legal),2) -- bowler's wicke
        AS wicket_pct
  FROM  v_ball                                       -- one row = one 
  WHERE TRIM(COALESCE(bowler_type,'')) <> ''         -- drop the blanks
        AND batsman_type IS NOT NULL                 -- and this too
  GROUP BY arm, batsman_type; 
""")
```

```python
q("""
--which spinner is the best wicket-taker?
-- bowler,wicket,wicket_perc  

SELECT bowler, SUM(bowler_wicket) AS wicket,
       ROUND(100.0 * SUM(bowler_wicket) / SUM(is_legal), 2) AS wicket_perc

FROM v_ball
WHERE bowler_type LIKE '%Legbreak%' OR
      bowler_type LIKE '%Offbreak%' OR
      bowler_type LIKE '%Orthodox%' OR
      bowler_type LIKE '%Wrist spin%' OR
      bowler_type LIKE '%Googly%'
GROUP BY bowler
ORDER BY wicket DESC
LIMIT 10;


""")
```

```python
q("""
--we need a death-overs bowler. who  cheapest in overs 16 to 20 ?
--bowler,balls.bowler_runs,economy

SELECT
    bowler,
    SUM(is_legal) AS balls,
    SUM(bowler_runs) AS bowler_runs,
    ROUND(
        6.0 * SUM(bowler_runs) / NULLIF(SUM(is_legal), 0),
        2
    ) AS economy

FROM v_ball
WHERE over_number BETWEEN 16 AND 20
GROUP BY bowler
HAVING balls > 300
ORDER BY economy ASC
LIMIT 10;

""")
```

```python
q("""
-- we need a powerplay strike bowler. who takes most wickets in overs 1 to 6 ?
-- bowler, balls,wicket,wicket_perc having balls > 300 

SELECT
    bowler,
    SUM(is_legal) AS balls,
    SUM(bowler_wicket) AS wicket,
    ROUND(
        100.0 * SUM(bowler_wicket) / NULLIF(SUM(is_legal), 0),
        2
    ) AS wicket_perc

FROM v_ball
WHERE over_number BETWEEN 0 AND 5
GROUP BY bowler
HAVING balls > 300
ORDER BY wicket DESC
LIMIT 10;

""")
```

```python
q("""
--which bowler has the best dot_ball rate at the death?
--bowler,ball,dot_balls,dot_perc 

SELECT
    bowler,
    SUM(is_legal) AS balls,
    SUM(CASE WHEN batter_runs = 0 THEN 1 ELSE 0 END) AS dot_balls,
    ROUND(
        100.0 * SUM(CASE WHEN batter_runs = 0 THEN 1 ELSE 0 END)
        / NULLIF(SUM(is_legal), 0),
        2
    ) AS dot_perc

FROM v_ball
WHERE phase = 'Death'
GROUP BY bowler
HAVING balls > 300
ORDER BY dot_perc DESC
LIMIT 10;

""")
```

```python
q("""
--Does the toss winner win the match more often than not?
--decisive_matches, toss winner , per

SELECT 
    COUNT(*) AS decisive_matches,
    SUM(
        CASE 
            WHEN toss_winner = match_winner THEN 1
            ELSE 0
        END
    ) AS toss_winner_won,
    ROUND(
        100.0 * SUM(
            CASE 
                WHEN toss_winner = match_winner THEN 1
                ELSE 0
            END
            
        )/count(*),2) AS pct
FROM matches_clean
WHERE result = 'win';
""")
```

```python
q(""" 
--which team relies most on boundaries for its run?
--run rate 100* batter_runs/legal_ball 4 6 only having sum(is_legal)>=10000
SELECT
    batting_team AS team,
    SUM(CASE WHEN batter_runs IN (4, 6) THEN batter_runs ELSE 0 END) AS boundary_runs,
    SUM(total_runs) AS total_runs,
    ROUND(
        100.0 * SUM(CASE WHEN batter_runs IN (4, 6) THEN batter_runs ELSE 0 END)
        / SUM(batter_runs), 2
    ) AS boundary_percentage
FROM v_ball
GROUP BY batting_team
HAVING SUM(is_legal) >=10000
ORDER BY boundary_percentage DESC
LIMIT 10;
""")
```

```python
q(""" 
--How does thw way batters get out change across the phases? 
--phases,wiket_kind=caught,bowled,run out,n
--v_ball table is_wicket 
SELECT
    phase,
    wicket_kind,
    COUNT(*) AS N,
    ROUND(
        100.0 * COUNT(*) /
        SUM(COUNT(*)) OVER (PARTITION BY phase),
        2
    ) AS percentage
FROM v_ball
WHERE is_wicket = 1
  AND wicket_kind IN ('caught', 'bowled', 'run out')
GROUP BY phase, wicket_kind
ORDER BY phase,N DESC;
""")
```
